## Creating DataFrames

Pandas DataFrames are one of the most powerful and widely used data structures in data analysis and manipulation tasks within Python. A DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

`pd.read_csv(filepath)`: Reads data from a CSV file into a DataFrame.

`pd.read_excel(filepath)`: Reads data from an Excel file into a DataFrame.

`pd.DataFrame(data)`: Creates a DataFrame from a variety of input data such as lists, dicts, or NumPy arrays.
- Example: `pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})`


## Examining DataFrames
`df.head(n)`: (DataFrame) returns the first `n` rows of the `df`. 

`df.tail(n)`: (DataFrame) Returns the last n rows of the `df`.

`df.columns.to_list()`: (List) returns a list of the column labels in `df`.

`len(df)`: (int) returns the number of rows in a `df`.

`len(clinical_data.columns)`: (int) returns the number of columns in a `df`.

`df.shape()`: (Tuple) returns a tuple of the (num_rows, num_cols) in a `df`.

`df.describe()`: (DataFrame) returns summary statistics for columns with quantitative variables in `df`.


## Indexing

`df.set_index('column')`: (DataFrame) Sets the DataFrame index using an existing column.

`df.reset_index()`: (DataFrame) Resets the index to the default integer index.

`df.index`: (Index) Returns the index values of the DataFrame as an Index data type.
- Use `df.index.to_list()` to convert the Index data type to a List.

#### Columns

**Label-based**

`df['column_label']`: (Series) returns a Series containing the values from `'column_label'` column in the `df`.

``df[column_labels_list]``: (DataFrame) returns a DataFrame containing only the columns in the `column_labels_list` (list).

`df.drop(column_labels_list, axis=1)`: (DataFrame) returns a DataFrame without the columns in the `column_labels_list` (list).

**Index-based**

`df.iloc[:, column_index]`: (Series) returns a Series containing the values from `column_index` column in the `df`.

`df.iloc[:, column_indices_list]`: (DataFrame) returns a DataFrame containing only the columns in the `column_indices_list` (list). 
- Note, you can also use slicing, e.g `df.iloc[:, 2:6]` will give you a DataFrame with column indices 2-5 (inclusive).

#### Rows
**Label-based**

`df.loc['row_label']`: (Series) returns a Series containing the values from the row with the label `'row_label'` in the df.

`df.loc[row_labels_list]`: (DataFrame) returns a DataFrame containing only the rows in the `row_labels_list` (list).

`df.drop(row_labels_list, axis=0)`: (DataFrame) returns a DataFrame without the rows in the `row_labels_list` (list).

**Index-based**

`df.iloc[row_index]`: (Series) returns a Series containing the values from the row at `row_index` in the df.

`df.iloc[row_indices_list]`: (DataFrame) returns a DataFrame containing only the rows at the indices in the `row_indices_list` (list).
- Note, you can also use slicing, e.g., `df.iloc[2:6]` will give you a DataFrame with row indices 2-5 (inclusive).

#### Columns or Rows
`df.filter(items = row_labels_list, like = substring)`: (DataFrame) returns a subset of the DataFrame rows or columns with specific labels or labels containing a substring.
- Can also work with more complicated cases, see [official documentation](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.filter.html) for string manipulation cases.

## Missing Data
`df.dropna()`: (DataFrame) Drops all rows that contain null values.

`df.fillna(value)`: (DataFrame) Fills null values with a specified value.

`df.isna()`: (DataFrame) Returns a boolean DataFrame to indicate if values are NA.

## Filtering DataFrames

`df[condition]`: (DataFrame) Filters the DataFrame to include only rows where the condition is True. The `condition` should be a list/Series of Boolean values (see examples below).

`df[~(condition)]`: (DataFrame) Filters the DataFrame to include only rows where the condition is False.

`df[(condition1) & (condition2)]`: (DataFrame) Filters the DataFrame to include only rows where both `condition1` and `condition2` is True.

`df[(condition1) | (condition2)]`: (DataFrame) Filters the DataFrame to include only rows where either `condition1` or `condition2` is True.

**Examples of Boolean Series:**

`df['column_label'] == value`: (Series) Returns a Series of Boolean values indicating whether elements in a column are equal to `value`.

`df['column_label'] > value`: (Series) Returns a Series of Boolean values indicating whether elements in a column are greater than `value`.

`df['column_label'] < value`: (Series) Returns a Series of Boolean values indicating whether elements in a column are less than `value`.

`df['column_label'].between(x, y)`: (Series) Returns a Series of Boolean values indicating whether elements in a column are between `x` and `y` (inclusive).

`df[column].isin(values)`: (Series) Returns a Series of Boolean values indicating if elements in a column are found in a list of `values`.

## Manipulating DataFrames

`df['new_column_label'] = data`: Updates `df` with a new column containing values from `data` (list-like data type).

`df.loc[new_index] = data`: Updates `df` with a new row containing values from `data` (list-like data type).

`df.drop('column', axis=1)`: (DataFrame) Deletes a column from the DataFrame.

`df.drop_duplicates()`: (DataFrame) Removes duplicate rows from the DataFrame.

`df.rename(columns={'old_name': 'new_name'}, axis = 1)`: (DataFrame) Renames columns.

`df.sort_values('column', ascending=True)`: (DataFrame) Sorts the DataFrame by a specific column in ascending order (by default).

`df.groupby('column')`: (DataFrame) Groups the DataFrame by a column for aggregation.
- You can group multiple columns by replacing `column` with a list of column labels.
- Once the data is grouped, you can apply a function to each group independently. This function can be an aggregation (like sum, mean, etc.), a transformation, or a filtration.

`df.pivot_table(values='column_label', index='column_label', columns='column_label', aggfunc='function_name')`: (DataFrame) Creates a pivot table.
- Note: common functions like count, sum, min, max, etc. can be input as a *string* into `aggfunc`. You can also use the function variable names. For example, `aggfunc="mean"` and `aggfunc=np.mean` are equivalent.

#### Aggregation Functions for Group

An aggregation function in Pandas is used to compute a summary statistic (or statistics) about each group after the groupby or pivot operation. Aggregation functions take a set of values, perform a calculation, and return a single value as the result. This is a key part of data analysis, allowing for the summary and examination of key trends and patterns in the data.

Let `grouped_df` be `df.groupby('column')`:

`grouped_df.count()`: Counts the number of elements in a group.
- Note `df.value_counts('column_label')` returns a Series with equivalent information to `df.groupby('column').count()`.

`grouped_df.sum()`: Adds up all values in a group.

`grouped_df.mean()`: Computes the average of the values in a group.

`grouped_df.min()`: Finds the smallest value in each group.

`grouped_df.max()`: Finds the largest value in each group.

`grouped_df.count()`: Counts the number of non-null values in each group.

`grouped_df.std()`: Computes the standard deviation of the values in a group, indicating the spread of a set of values.

`grouped_df.var()`: Computes the variance of the values in each group.

`grouped_df.median()`: Finds the median (middle value) of the values in each group.

`grouped_df.quantile(q)`: Calculates the q-th quantile of the values in each group (e.g., the median is the 0.5-quantile).

Custom aggregation: 
`df.groupby('column').agg(custom_function)`: Applies a custom function to group or pivot data. The function should take a DataFrame or Series, perform the desired operation, and return a value or a Series.

#### Applying Functions
`df['column'].apply(function)`: (Series) applies a function on each value in a column. The custom function should have only one argument and one output.

`df.apply(custom_function, axis=1, param1=value1, param2=value2, ...)`: (Series) applies a function using multiple column values for arguments.

#### Merging DataFrames

`pd.merge(df1, df2, on='column')`: (DataFrame) Merges two DataFrames based on a join key.

`pd.concat([df1, df2], axis=0)`: (DataFrame) Concatenates DataFrames along a particular axis.

## Series

A Pandas Series is a one-dimensional array-like object that can hold many data types, including integers, floats, strings, Python objects, and more. It is one of the primary data structures in Pandas, alongside the DataFrame. A Series is similar to a column in a spreadsheet or a database table.

`pd.Series(data)`: (Series) Creates a new series containing `data` (list, ndarray, dict, or scalar).

`series.head(n)`: (Series) Returns the first n elements of the Series.

`series.tail(n)`: (Series) Returns the last n elements of the Series.

`series.sum()`: (Scalar) Returns the sum of the Series elements.

`series.mean()`: (Scalar) Returns the mean of the Series elements.

`series.median()`: (Scalar) Returns the median of the Series elements.

`series.mode()`: (Series) Returns the mode(s) of the Series.

`series.min()`: (Scalar) Returns the smallest element in the Series.

`series.max()`: (Scalar) Returns the largest element in the Series.

`series.std()`: (Scalar) Returns the standard deviation of the Series.

`series.describe()`: (DataFrame) Returns a summary of statistics of the Series.

`series.value_counts()`: (Series) Returns a Series containing counts of unique values.

`series.apply(func)`: (Series) Applies a function to each element in the Series.

`series.sort_values(ascending=True)`: (Series) Sorts the Series in ascending or descending order.

`series.str.len()`: (Series) Calculates the length of each string in the Series.

`series.unique()`: (Array) Returns the unique values in the Series.

### Visualizations

**Basic Plotting:**
   - **Line Plot (default)**: `df.plot()` for line graphs.
   - **Bar Plot**: `df.plot(kind='bar')` for vertical bar graphs, `df.plot(kind='barh')` for horizontal bar graphs.
   - **Histogram**: `df.plot(kind='hist')` for histograms.
   - **Box Plot**: `df.plot(kind='box')` for box plots.
   - **Scatter Plot**: `df.plot(kind='scatter', x='col1', y='col2')` to create a scatter plot.
   - **Pie Chart**: `df['column'].plot(kind='pie', autopct='%1.1f%%')` for pie charts. Use `autopct` to display percentage.
   - **Area Plot**: `df.plot(kind='area')` for area plots. Useful for stacked area plots.
Note, 1D plots can also be done with `series` instead of `df`.

**Customizing Plots:**
   - **Title**: Add `title='Your Title'` in the plot function.
   - **Axis Labels**: Add `xlabel='X-axis Label'`, `ylabel='Y-axis Label'`.
   - **Legend**: Add `legend=True` to display legend. Customize with `legend=['label1', 'label2']`.
   - **Color**: Use `color='color_name'` or `colors=['color1', 'color2']` for multiple columns.
   - **Size**: Use `figsize=(width, height)` to set the size of the plot.
   - **Grid**: Add `grid=True` to show grid lines.


**Note:** While basic plotting can be done with just Pandas, it is often recommended to use Matplotlib or Seaborn for more advanced plots.
