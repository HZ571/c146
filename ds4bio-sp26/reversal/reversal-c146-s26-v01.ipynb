```python
# Approach 2
s = "Definition of a college professor: someone who talks in other people's sleep." # W.H. Auden

the_mess = {random.random(): [idx, letter] for idx, letter in enumerate(s)}
r = ""
length = len(s)
count = length - 1
while count >= 0:
    for letter in the_mess:
        total = the_mess[letter]
        if total[0] == count:
            r += total[1]
            break
    count -= 1
r = "".join(list(r)[::-1][::-1])

print ("String  :",s,"\nReversed:",r)
```
